## 1. Load the tools

```python
import io
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
from IPython.display import display
from google.colab import files
```

## 2. Upload your original CSV
Run this cell, click Choose Files, and select one CSV from your computer.

```python
uploaded = files.upload()
if len(uploaded) != 1:
    raise ValueError('Please upload exactly one CSV and run this cell again.')
filename = next(iter(uploaded))
df = pd.read_csv(io.BytesIO(uploaded[filename]))
print('Uploaded:', filename)
```

## 3. Inspect the file
Look at the column names and first rows. Decide what one row means before selecting X and Y.

```python
print('Rows:', len(df))
print('Columns:', list(df.columns))
display(df.head())
```

## 4. Choose your variables
Edit the two names below. Tips: total_bill → tip. Cars: weight → mpg. Penguins: flipper_length_mm → body_mass_g. Use exactly one predictor. For your own file, enter your numerical column names.

```python
x_column = 'total_bill'
y_column = 'tip'
if x_column == y_column:
    raise ValueError('Choose different columns for X and Y.')
data = df[[x_column, y_column]].apply(pd.to_numeric, errors='coerce')
data = data.replace([np.inf, -np.inf], np.nan).dropna()
print('Original rows:', len(df))
print('Usable rows:', len(data))
print('Rows removed for selected X/Y:', len(df) - len(data))
if len(data) < 3 or data[x_column].nunique() < 2:
    raise ValueError('Choose data with at least three usable rows and varying X.')
if data[y_column].nunique() < 2:
    raise ValueError('Choose a Y column that varies across rows.')
print('Observed X range:', data[x_column].min(), 'to', data[x_column].max())
```

## 5. Fit one straight line and read the tables
The first table gives the intercept and slope. The second gives the sample size and R-squared. Use these in your explanation.

```python
X = sm.add_constant(data[[x_column]], has_constant='add')
model = sm.OLS(data[y_column], X).fit()
intercept, slope = model.params.iloc[0], model.params.iloc[1]
coefficient_table = pd.DataFrame({'Term': ['Intercept', x_column], 'Coefficient': [intercept, slope]})
model_info = pd.DataFrame({'Observations': [int(model.nobs)], 'R-squared': [model.rsquared]})
display(coefficient_table.round(3))
display(model_info.round(3))
print(f'Predicted {y_column} = {intercept:.3f} + ({slope:.3f}) * {x_column}')
coefficient_table.to_csv('regression_coefficients.csv', index=False)
model_info.to_csv('regression_model_info.csv', index=False)
```

## 6. Draw the fitted line
The dots are observations. The straight line represents predicted Y. Add the variable units to the axis labels below.

```python
x_axis_label = x_column + ' (add units here)'
y_axis_label = y_column + ' (add units here)'
x_line = np.linspace(data[x_column].min(), data[x_column].max(), 100)
y_line = intercept + slope * x_line
plt.figure(figsize=(8, 5))
plt.scatter(data[x_column], data[y_column], alpha=0.65, label='Observed data')
plt.plot(x_line, y_line, color='darkred', linewidth=2, label='Fitted straight line')
plt.xlabel(x_axis_label)
plt.ylabel(y_axis_label)
plt.title('Simple linear regression')
plt.legend()
plt.tight_layout()
plt.savefig('regression_plot.png', dpi=180)
plt.show()
```

## 7. Make one prediction
Replace None with an X value inside the observed range. For tips.csv, you could use 20. Explain this prediction with its units.

```python
x_to_predict = None
if x_to_predict is None:
    print('Enter your chosen X value above, then run this cell again.')
else:
    if not data[x_column].min() <= x_to_predict <= data[x_column].max():
        raise ValueError('Choose an X value inside the observed range.')
    print(f'At {x_column} = {x_to_predict}, predicted {y_column} = {intercept + slope * x_to_predict:.3f}')
```
